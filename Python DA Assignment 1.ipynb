```python
import numpy as np
import pandas as pd
```

**Numpy Operations**

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# Conversion 
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_f)
#Find min,max,mean
print("\nMaximum temperature:", np.max(temperatures_w1))
print("Minimum temperature:", np.min(temperatures_w1))
print("Mean temperature:", np.mean(temperatures_w1))
```

```python
#Array slicng and indexing
print("First three days:", temperatures_w1[:3])
print("\nWeekend temperatures:", temperatures_w1[-2:])
print("\nMiddle three days:", temperatures_w1[2:5])
```

```python
#Creation of 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

```python
#Inspect and Slice
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)
# Extract temperatures
print("\nWeek 1:", temperatures[0])
print("Week 2:", temperatures[1])
# Extract weekend temperature
print("\nWeekend temperatures for both weeks:")
print(temperatures[:, -2:])
```

**PandasSeries**

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
#Indexing and Slicing

print("1st rank mark:", marks.iloc[0])
print("\nTop 3 ranks:")

print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\n3rd rank mark:", marks.iloc[2])

print("\nMarks greater than 90:")
print(marks[marks > 90])
```

```python
# Manipulating Series

marks.loc['Rank1'] = 100
print("After modifying Rank1:")
print(marks)

# Remove the last rank
marks = marks.drop('Rank5')
print("\nAfter removing Rank5:")
print(marks)


cgpa = marks / 10
print("\nCGPA:")
print(cgpa)
```

**Pandas DataFrame**

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
print("Transactions DataFrame:")
display(transactions)

print("\nHead:")
display(transactions.head())

print("\nTail:")
display(transactions.tail())

print("\nShape:", transactions.shape)
print("Column names:", transactions.columns)
print("Data types:")
print(transactions.dtypes)

print("\nDataFrame information:")
transactions.info()
```

```python

print(transactions[['ProductCategory', 'Amount']])

print("\nLast three columns:")
print(transactions.iloc[:, -3:])

north_transactions = transactions[
    (transactions['Region'] == 'North') & (transactions['Amount'] > 200)
]
print("\nNorth transactions with Amount greater than 200:")
print(north_transactions)

print("\nProduct category counts:")
print(transactions['ProductCategory'].value_counts())

print("\nUnique regions:")
print(transactions['Region'].unique())

print("\nMean amount by region:")
print(transactions.groupby('Region')['Amount'].mean())
```

**Manipulating the DataFrame**

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("After modifying TransactionID 102:")
print(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter adding Discount:")
print(transactions)

transactions = transactions[transactions['TransactionID'] != 109]
print("\nAfter removing TransactionID 109:")
print(transactions)

transactions = transactions.drop(columns=['Discount'])
print("\nFinal DataFrame:")
print(transactions)
```
